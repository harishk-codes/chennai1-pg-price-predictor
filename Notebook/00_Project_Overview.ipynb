## 1. Project Background

Paying Guest (PG) accommodation is widely used by students and working professionals in Chennai. The monthly cost of a PG can vary considerably even between properties located in the same broad area.

The purpose of this project is to study the factors associated with PG rental prices and then use those factors to build a machine learning model for rent prediction.

The analysis focuses on information such as:

- Location
- Room-sharing arrangement
- Intended occupants
- Parking
- Food availability
- Other amenities
- Rental price
- Security deposit

A useful prediction system could help tenants compare listings more easily and could also help property owners understand how different property characteristics relate to rent.


## 2. Problem Definition

The main problem is to estimate the monthly rent of a PG from the information available for that listing.

### Input

The model can use characteristics such as:

- Locality
- Occupancy type
- Gender category
- Parking availability
- Food availability
- Amenities
- Location-related scores
- Deposit

### Output

The target variable is:

**Monthly PG rent**

This is a supervised machine learning regression problem because the value being predicted is numeric.


## 3. Project Objectives

The project has the following objectives:

1. Understand the structure and quality of the PG listing data.
2. Identify missing, duplicated, invalid, and unnecessary records or columns.
3. Clean the dataset without removing useful information unnecessarily.
4. Explore individual variables and their distributions.
5. Study how different characteristics are related to rent.
6. Examine whether combinations of features produce different pricing patterns.
7. Prepare a reliable dataset for machine learning.
8. Train and compare regression models in the later stages.
9. Evaluate how accurately the final model predicts unseen PG rents.


## 4. Dataset Description

Each record represents a room type associated with a PG property. Therefore, a single property can appear in more than one row when it offers different sharing arrangements.

## 6. EDA Methodology

Exploratory Data Analysis (EDA) is performed before model training so that the dataset is understood properly.

The analysis is divided into five stages:

| Stage | Purpose |
|---|---|
| Stage 1 | Inspect the raw data and identify issues |
| Stage 2 | Apply the required cleaning decisions |
| Stage 3 | Study variables individually |
| Stage 4 | Examine relationships between features and rent |
| Stage 5 | Study multiple features together |

This order is useful because later analysis depends on the quality of the cleaned data.


## 7. Stage 1 — Initial Data Inspection

### Goal

The first step is to understand what is present in the raw dataset before making changes.

The following checks should be performed:

- Number of rows and columns
- Column names
- Data types
- Missing values
- Duplicate records
- Unusual values
- Numeric ranges
- Low-information columns
- Potential identifier or free-text columns

### Missing data

Missing values should not automatically be treated in the same way.

The reference analysis identified:

- Missing transit and lifestyle scores
- Missing amenity information
- A small number of rows missing important fields such as rent, deposit, or room type
- A highly incomplete `gate_closing_time` column

The meaning of a missing value must be considered before choosing an imputation or removal strategy.

### Special values

The reference dataset used `-10` in the transit score field as a sentinel value for unavailable data. It therefore should not be interpreted as a genuine negative score.

The analysis also identified very low rent values as potential data-entry errors.

### Duplicate records

Duplicate listings should be investigated before removal. An exact duplicate may represent repeated data, while similar records can sometimes represent genuinely different room types.


## 8. Stage 2 — Data Quality and Cleaning

### Goal

Apply the decisions made during the initial inspection and create a consistent dataset for analysis.

The reference workflow included the following actions:

- Remove identifier and free-text fields that do not provide useful predictive information.
- Remove columns with very high missingness when their contribution is limited.
- Remove records without essential target or room information.
- Remove clearly invalid low-rent records.
- Remove duplicate property/room combinations.
- Standardize amenity fields as Boolean values.
- Replace the transit-score sentinel value with a missing marker.
- Fill missing location-based scores using locality-level information, with a broader fallback where required.
- Retain missingness indicators when the fact that a value was unavailable may itself carry information.

The reference cleaning process reduced the data from 1,783 records to 1,471 records before one additional confirmed data error was removed.

The resulting reference dataset contained 1,470 rows and 28 columns.

**For a new dataset, these counts must be recalculated rather than copied.**


## 9. Stage 3 — Individual Feature Analysis

At this stage, each variable is considered separately.

### Numerical variables

Useful checks include:

- Minimum
- Maximum
- Mean
- Median
- Standard deviation
- Distribution
- Skewness
- Outliers

The reference analysis found that rent was strongly right-skewed, with many listings concentrated in the lower and middle price range and a smaller number of expensive listings.

The deposit variable was also highly skewed.

### Categorical variables

For categorical features, inspect:

- Number of categories
- Frequency of each category
- Percentage distribution
- Rare categories
- Possible class imbalance

Important examples include locality, occupancy, gender, and intended occupant category.

### Boolean variables

Amenity fields can be summarized using the proportion of listings where each facility is available.

Highly redundant fields should also be checked. In the reference data, meal-specific fields were closely aligned with the overall food-included field.


## 10. Stage 4 — Relationship with Rent

The next question is not simply what each feature looks like, but whether it is associated with the target variable.

### Numeric features

Examples:

- Deposit vs rent
- Transit score vs rent
- Lifestyle score vs rent
- Latitude vs rent
- Longitude vs rent

Correlation can provide an initial indication of linear association, but it should not be treated as proof of causation.

### Categorical features

Compare rent distributions across:

- Occupancy types
- Localities
- Gender categories
- Parking types
- Food availability
- Amenity availability

The reference analysis identified occupancy and locality as important pricing-related variables. It also found a positive relationship between deposit and rent.

The exact effect sizes and averages must be recalculated when using a different dataset.


## 11. Stage 5 — Combined Feature Analysis

Individual feature analysis is not always enough.

For example, a parking option may appear to increase rent simply because properties offering that option contain more expensive room types.

Similarly, a difference between gender categories may be related to differences in the type of customers served by those properties.

### Important interaction to investigate

**Locality × Occupancy**

The reference analysis found that the price difference between single and shared rooms was not the same across all localities.

This means a useful prediction model should be able to represent combinations of features rather than assuming that every feature has exactly the same effect everywhere.

### Other useful comparisons

- Parking × Occupancy
- Parking × Locality
- Food × Occupancy
- Gender × Intended occupant category
- Amenities × Occupancy

The goal is to distinguish a genuine relationship from a relationship caused by another variable.


## 12. Main Observations

Based on the supplied reference analysis, the major areas that deserve attention in the machine learning stage are:

- Room-sharing arrangement is strongly associated with rent.
- Locality has an important relationship with pricing.
- Deposit and rent move together to a noticeable degree.
- Car parking showed a premium in the reference analysis even after accounting for location and room type.
- Food inclusion showed a smaller positive difference in the reference analysis.
- Transit and lifestyle scores showed weaker direct relationships with rent.
- Some categorical variables may look important initially but become weaker after another related variable is considered.
- Locality and occupancy can interact, so they should not always be treated as completely independent effects.

These observations are hypotheses from the reference analysis and should be verified again using the actual dataset used for the final model.

## 14. Machine Learning Preparation

After EDA and cleaning, the dataset needs to be converted into a form suitable for machine learning.

### Category encoding

Text-based categories such as locality, occupancy, gender, and parking need numerical representations.

Different encoding methods can be considered depending on the feature.

### Skewed numerical variables

Rent and deposit may contain long right tails. A logarithmic transformation such as `log1p` can be considered when appropriate.

### Train, validation, and test sets

The cleaned data should be separated into:

- Training data — used to fit the model
- Validation data — used during model selection and tuning
- Test data — reserved for final evaluation

### Candidate models

Possible regression models include:

- Linear Regression
- Random Forest Regressor
- Gradient Boosting
- XGBoost / LightGBM
- CatBoost

The final model should be selected using validation performance rather than assumption.

### Evaluation

Useful regression metrics include:

- MAE
- RMSE
- R²

Feature importance or other explainability methods can then be used to compare model behavior with the patterns discovered during EDA.

## 16. Glossary

**EDA (Exploratory Data Analysis)**  
The process of examining data before modelling to understand its structure, quality, distributions, and relationships.

**Feature**  
An input variable used by a machine learning model.

**Target variable**  
The value the model is trained to predict. Here, the target is monthly rent.

**Univariate analysis**  
Studying one variable at a time.

**Bivariate analysis**  
Studying the relationship between two variables.

**Multivariate analysis**  
Studying several variables together.

**Correlation**  
A numerical measure describing how two variables move together. It does not by itself establish causation.

**Imputation**  
Filling missing information using a defined estimation strategy.

**Outlier**  
A value that is unusually far from the rest of the observations.

**Skewness**  
A measure describing the asymmetry of a distribution.

**Encoding**  
Converting categorical information into a numerical representation suitable for a model.

**Interaction effect**  
A situation where the effect of one feature changes depending on another feature.

**Multicollinearity**  
A condition where input features are strongly related to each other.

**Overfitting**  
When a model learns the training data too closely and performs poorly on unseen data.

**Sentinel value**  
A special value used to indicate a condition such as unavailable data rather than a real measurement.

**PG (Paying Guest)**  
A form of rented accommodation in which a person pays for a room, often with shared facilities or included amenities.
