```python
# 1.2 Missing values
def missing_info(df):
    missing_info = pd.DataFrame({
        'missing count': df.isna().sum(),
        'missing %': (df.isna().mean() * 100).round(2),
        'Data type': df.dtypes,
    })

    missing_info = missing_info[missing_info['missing count'] > 0]
    missing_info.sort_values('missing %', ascending = False)
    return missing_info

missing_info(df)
```

```python
# 1.3 Duplicates
def isDuplicates(df: pd.DataFrame):
    print(f'Total duplicates: {df.duplicated().sum()}')
    return df[df.duplicated(keep=False)].sort_values(by=df.columns.tolist()).head(20)

isDuplicates(df)
```

# EDA Phase 1 — Sanity Check

## Overview

The purpose of this phase is to perform an initial health check of the raw Chennai PG dataset before applying data-cleaning operations.

The analysis focuses on:

- Dataset structure and data types
- Missing values
- Duplicate records
- Categorical variables
- Basic descriptive statistics

The findings from these checks are then used to decide what should be retained, transformed, or removed in the cleaning stage.


## 1. Dataset Structure and Data Types

The first step is to inspect the dataset using `df.info()`.

The check helps identify:

- Number of records and columns
- Data types
- Columns containing missing values
- Boolean fields stored as object/text values

Some amenity fields such as `wifi` and `laundry` are stored as `object` values instead of Boolean values.

### Decision

Useful columns are retained, while incorrectly typed Boolean fields will be converted into an appropriate format during cleaning.


## 2. Missing Value Analysis

Missing values are examined for every column.

### Score fields

`transit_score` and `lifestyle_score` contain a large proportion of missing values, approximately half of the records.

These missing values are treated as unavailable information rather than valid scores.

### Decision

For these fields:

1. Create a missingness indicator.
2. Fill missing values using the median score for the same locality.
3. Use the overall median as a fallback when a locality does not have an available score.

### Amenity fields

Several amenity columns contain substantial missing information.

The interpretation used in the analysis is:

| Value | Interpretation |
|---|---|
| `True` | Amenity is available |
| `False` | Amenity is unavailable |
| `NaN` | Information was not supplied |

For modelling, missing amenity values are treated as unavailable.

### Critical fields

A small proportion of records have missing values in important fields such as `rent`.

Since rent is the target variable, missing target values are not imputed and those records are removed.

### `gate_closing_time`

This field has very high missingness and limited usefulness for rent prediction.

**Decision:** Remove this column.


## 3. Duplicate Record Check

Duplicate records are checked before further analysis.

The same PG property may appear more than once because different occupancy options can exist for the same property.

Therefore, duplicate handling should consider the property identifier together with occupancy rather than relying only on the complete row.

The cleaning rule used in the analysis keeps the first record for each:

`id + occupancy`

combination.


## 4. Categorical Feature Check

The unique values of categorical columns are examined to understand their structure.

Important observations include:

- `id` is an identifier and is not useful as a predictive feature.
- `title` is high-cardinality free text.
- `address` is also high-cardinality, while `locality` provides a more useful geographic representation.
- `locality` contains 36 categories.
- `gender` contains three categories.
- `available_for` contains three categories.
- `occupancy` contains four main categories, with some missing values.
- Amenity fields contain Boolean/missing values.
- `parking` contains different parking combinations.

### Decision

Identifier and unnecessary free-text fields should be removed, while useful categorical variables such as locality, occupancy, gender, and parking should be retained for later encoding.


## 5. Basic Statistical Summary

Descriptive statistics are generated separately for numerical and categorical variables.

For numerical columns, the analysis examines values such as:

- Count
- Mean
- Standard deviation
- Minimum
- Maximum
- Quartiles

For categorical columns, the analysis examines:

- Count
- Number of unique categories
- Most frequent category
- Frequency of the most common category

This provides an initial understanding of the dataset before cleaning and modelling.
